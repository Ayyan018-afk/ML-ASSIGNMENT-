```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
data = pd.read_excel("std_data.xlsx")
```

```python
data
```

```python
data.columns = data.iloc[0]
data = data[1:]
data.reset_index(drop=True, inplace=True)
```

```python
cols = ['Study_Hours_Per_Day','Attendance_%','Previous_GPA',
        'Assignments_Submitted','Sleep_Hours','Final_Marks']

for col in cols:
    data.loc[:,col] = pd.to_numeric(data[col], errors='coerce')
```

```python
data.duplicated().sum()
```

```python
data = data.drop_duplicates()
```

```python
data = data.fillna(data.mean(numeric_only = True))
```

```python
data.isna().sum()
```

```python
data.head(1)
```

```python
X = data.drop(columns=['Final_Marks','Student_ID','Student_Name'])
y = data['Final_Marks']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer( transformers=[('num', StandardScaler(), X.columns)])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LinearRegression

pipeline = Pipeline(steps=[('preprocessing', preprocessor),('model', LinearRegression())])
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.58, random_state=86)
```

```python
pipeline.fit(X_train, y_train)
```

```python
y_pred = pipeline.predict(X_test)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test, y_pred)
print(r2)
```
